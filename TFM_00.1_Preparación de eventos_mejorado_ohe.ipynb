```python
df = pd.read_csv('data/eventos_extra_raw.csv')
```

```python
df.info()
```

```python
df['Fecha'] = pd.to_datetime(df['Fecha'], format= "%d/%m/%Y")
```

```python
df.head(25)
```

```python
df['Nombre festivo'].value_counts()
```

```python
df = df.sort_values(by='Fecha')
df.reset_index(drop= True, inplace= True)
```

```python
nombres_nuevos = {
    'Año Nuevo': 'NewYear',
    'Día de Martin Luther King Jr.': 'MartinLutherKing',
    'Año Nuevo Chino': 'ChineeseNewYear',
    'Superbowl': 'Superbowl',
    'Día de los Presidentes': 'President',
    'Día de San Patricio': 'StPatrick',
    'Día de los Caídos': 'Memorial',
    'Vesak': 'Buddhist',
    'Juneteenth': 'Juneteenth',
    'Día de la Independencia': 'Independence',
    'Ramadán': 'RamadanStarts',
    'Eid al-Fitr': 'RamadanEnds',
    'Día del Trabajo': 'Labour',
    'Rosh Hashaná': 'JewNewYear',
    'Yom Kipur': 'YomKipur',
    'Día de Colón': 'Columbus',
    'Diwali': 'Hindu',
    'Día de los Veteranos': 'Veterans',
    'Día de Acción de Gracias': 'Thanksgiving',
    'Navidad': 'Christmas'
}

df['Nombre festivo'] = df['Nombre festivo'].replace(nombres_nuevos)
```

```python
nombres_grupo = {
    'Nacional': 'national', 
    'Chino': 'chinese', 
    'Budismo': 'buddhist', 
    'Islam': 'islam', 
    'Judaísmo': 'jew', 
    'Hinduismo': 'hindu'
}

df['Grupo'] = df['Grupo'].replace(nombres_grupo)
```

```python
df.head(25)
```

```python
# Creamos yearweek para usarlo como key en el merge

df['yearweek'] = df['Fecha'].dt.isocalendar().year.astype(str) + \
df['Fecha'].dt.isocalendar().week.astype(str).str.zfill(2)
```

```python
df.head(20)
```

```python
df[df['yearweek'].duplicated(keep=False)]
```

Como tenemos eventos que caen dentro de la misma semana, decidimos un enfoque de One Hot Encoding para no perder información

```python
# Eliminamos las columnas que ya no usaremos
df.drop(['Fecha', 'Descripción'], axis=1, inplace=True)
```

```python
# Cambiamos el nombre de las columnas para que sean los prefijos de las columnas OHE
df.rename(columns={'Nombre festivo': 'event', 'Grupo': 'event_group'}, inplace=True)

df_ohe = pd.get_dummies(df, columns = ['event','event_group'])
```

```python
df_ohe.head()
```

```python
list(df_ohe.columns)
```

```python
# Creamos una columna extra que representará si hay evento en esa semana. Es True para todas las semanas de este df
df_ohe['has_event'] = True
```

```python
# Codificamos con 1 y 0
for col in list(df_ohe.columns):
    df_ohe[col] = np.where(df_ohe[col] == True, 1, df_ohe[col])
    df_ohe[col] = np.where(df_ohe[col] == False, 0, df_ohe[col])
```

```python
df_ohe
```

```python
# Comprobamos que hay varias filas con el mismo yearweek. Son semanas con más de un evento
df_ohe[df_ohe['yearweek'].duplicated(keep= False)]
```

```python
# Agrupamos las semanas con más de un evento en una única semana
df_ohe_agrupado = df_ohe.groupby('yearweek', as_index=False).max()
```

```python
df_ohe_agrupado[df_ohe_agrupado['yearweek'].duplicated(keep= False)]
```

```python
df_ohe_agrupado.to_csv('data/eventos_mejorados_ohe.csv')
```
